```python
import logging
import sys
import uuid


def configure_logging(
    level: int = logging.INFO,
) -> logging.Logger:
    """
    Configure application-wide logging.

    A request ID can later be attached to log records so that
    API requests, worker jobs and AI calls can be correlated.
    """

    logger = logging.getLogger("ai_study_companion")

    if logger.handlers:
        return logger

    logger.setLevel(level)

    handler = logging.StreamHandler(sys.stdout)

    formatter = logging.Formatter(
        "%(asctime)s | "
        "%(levelname)s | "
        "%(name)s | "
        "%(message)s"
    )

    handler.setFormatter(formatter)
    logger.addHandler(handler)

    return logger


def create_request_id() -> str:
    """Create a unique ID for a request or AI operation."""
    return str(uuid.uuid4())


def log_event(
    logger: logging.Logger,
    event: str,
    **fields,
) -> None:
    """Write an event with structured key/value information."""

    details = " ".join(
        f"{key}={value}"
        for key, value in fields.items()
    )

    if details:
        logger.info("%s | %s", event, details)
    else:
        logger.info("%s", event)
```
