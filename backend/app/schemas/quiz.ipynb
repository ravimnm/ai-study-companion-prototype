```python
from typing import Literal

from pydantic import BaseModel, Field


class QuizGenerateRequest(BaseModel):
    project_id: str
    concept_ids: list[str] = Field(default_factory=list)
    question_count: int = Field(default=5, ge=1, le=20)
    difficulty: Literal["adaptive", "easy", "medium", "hard"] = "adaptive"


class QuizStartRequest(BaseModel):
    assessment_id: str


class AnswerSubmission(BaseModel):
    question_id: str
    answer: str = Field(min_length=1, max_length=5000)


class OpenEndedEvaluationResponse(BaseModel):
    overall_score: float = Field(ge=0.0, le=1.0)
    understanding: Literal[
        "poor",
        "partial",
        "good",
        "excellent"
    ]
    correct_concepts: list[str] = Field(default_factory=list)
    missing_concepts: list[str] = Field(default_factory=list)
    misconceptions: list[str] = Field(default_factory=list)
    feedback: str
    mastery_signal: float = Field(ge=0.0, le=1.0)


class QuizQuestionResponse(BaseModel):
    id: str
    question: str
    type: Literal["mcq", "open_ended"]
    concept_id: str
    difficulty: Literal["easy", "medium", "hard"]
    options: list[str] = Field(default_factory=list)
    explanation: str | None = None


class QuizResponse(BaseModel):
    id: str
    project_id: str
    title: str
    status: str
    questions: list[QuizQuestionResponse] = Field(default_factory=list)
```
