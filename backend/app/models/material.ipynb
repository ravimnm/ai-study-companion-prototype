```python
from datetime import datetime, timezone
from typing import Literal

from pydantic import BaseModel, Field


class Material(BaseModel):
    """
    MongoDB document representing uploaded learning material.

    The actual PDF binary should not be stored directly in this
    document. file_path/storage reference points to the stored file.
    """

    id: str | None = None

    user_id: str
    project_id: str

    title: str = Field(
        min_length=1,
        max_length=200,
    )

    file_name: str
    file_type: str = "application/pdf"

    file_path: str | None = None

    status: Literal[
        "pending",
        "processing",
        "completed",
        "failed",
    ] = "pending"

    page_count: int | None = None
    chunk_count: int = 0
    concept_count: int = 0

    error_message: str | None = None

    processing_attempts: int = 0

    created_at: datetime = Field(
        default_factory=lambda: datetime.now(timezone.utc)
    )

    updated_at: datetime = Field(
        default_factory=lambda: datetime.now(timezone.utc)
    )
```
