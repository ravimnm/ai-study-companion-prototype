```python
from datetime import datetime, timezone

from pydantic import BaseModel, Field


class Project(BaseModel):
    """
    MongoDB document representing a learning Project.

    A Project is the primary isolation boundary for learning
    context, materials, conversations, assessments and mastery.
    """

    id: str | None = None

    user_id: str
    space_id: str

    name: str = Field(
        min_length=1,
        max_length=150,
    )

    description: str | None = Field(
        default=None,
        max_length=1000,
    )

    goal: str | None = Field(
        default=None,
        max_length=1000,
    )

    created_at: datetime = Field(
        default_factory=lambda: datetime.now(timezone.utc)
    )

    updated_at: datetime = Field(
        default_factory=lambda: datetime.now(timezone.utc)
    )
```
