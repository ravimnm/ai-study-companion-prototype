```python
from datetime import datetime, timezone
from typing import Literal
from pydantic import BaseModel, Field, field_validator
```

```python
class Mastery(BaseModel):
    id: str | None = None

    user_id: str
    project_id: str
    concept_id: str

    # Mastery score represented as a value between 0 and 1
    score: float = 0.0

    # Confidence in the estimated mastery score
    confidence: float = 0.0

    trend: Literal[
        "improving",
        "stable",
        "needs_attention"
    ] = "stable"

    assessment_count: int = 0
    last_assessed_at: datetime | None = None

    # Useful for explaining why mastery changed
    recent_evidence: list[str] = Field(default_factory=list)

    created_at: datetime = Field(
        default_factory=lambda: datetime.now(timezone.utc)
    )

    updated_at: datetime = Field(
        default_factory=lambda: datetime.now(timezone.utc)
    )

    @field_validator("score", "confidence")
    @classmethod
    def validate_probability(cls, value: float) -> float:
        if not 0.0 <= value <= 1.0:
            raise ValueError("Value must be between 0 and 1")
        return value

    @field_validator("assessment_count")
    @classmethod
    def validate_assessment_count(cls, value: int) -> int:
        if value < 0:
            raise ValueError("assessment_count cannot be negative")
        return value


print("Model: Mastery")
print("Scope: User + Project + Concept")
print("Score range: 0.0 - 1.0")
print("Trend values: improving, stable, needs_attention")
print("Purpose: Track concept-level learning progress")
```
