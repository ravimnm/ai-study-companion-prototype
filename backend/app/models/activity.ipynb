```python
from datetime import datetime, timezone
from typing import Any
from pydantic import BaseModel, Field
```

```python
class Activity(BaseModel):
    id: str | None = None

    user_id: str
    project_id: str | None = None

    event_type: str
    description: str | None = None

    entity_type: str | None = None
    entity_id: str | None = None

    metadata: dict[str, Any] = Field(default_factory=dict)

    created_at: datetime = Field(
        default_factory=lambda: datetime.now(timezone.utc)
    )


print("Model: Activity")
print("Scope: User + optional Project")
print("Purpose: Store learning and system events")
print("Example events:")
print("- PROJECT_CREATED")
print("- MATERIAL_UPLOADED")
print("- TUTOR_MESSAGE_SENT")
print("- QUIZ_COMPLETED")
print("- MASTERY_UPDATED")
print("- RECOMMENDATION_CREATED")
```
