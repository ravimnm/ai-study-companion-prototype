```python
from datetime import datetime, timezone
from typing import Literal
from pydantic import BaseModel, Field
```

```python
class AssessmentQuestion(BaseModel):
    id: str
    question: str
    type: Literal["mcq", "open_ended"]
    concept_id: str
    difficulty: Literal["easy", "medium", "hard"]
    options: list[str] = Field(default_factory=list)
    correct_option: str | None = None
    expected_concepts: list[str] = Field(default_factory=list)
    explanation: str | None = None
    source_chunk_ids: list[str] = Field(default_factory=list)


class AssessmentAnswer(BaseModel):
    question_id: str
    answer: str
    is_correct: bool | None = None
    score: float | None = None
    feedback: str | None = None
    missing_concepts: list[str] = Field(default_factory=list)
    misconceptions: list[str] = Field(default_factory=list)
    answered_at: datetime = Field(
        default_factory=lambda: datetime.now(timezone.utc)
    )


class Assessment(BaseModel):
    id: str | None = None
    project_id: str
    user_id: str
    title: str
    assessment_type: Literal["adaptive_quiz", "practice", "assessment"]
    status: Literal["draft", "in_progress", "completed", "failed"] = "draft"
    questions: list[AssessmentQuestion] = Field(default_factory=list)
    answers: list[AssessmentAnswer] = Field(default_factory=list)
    score: float | None = None
    max_score: float | None = None
    started_at: datetime | None = None
    completed_at: datetime | None = None
    created_at: datetime = Field(
        default_factory=lambda: datetime.now(timezone.utc)
    )
    updated_at: datetime = Field(
        default_factory=lambda: datetime.now(timezone.utc)
    )


print("Model: Assessment")
print("Nested model: AssessmentQuestion")
print("Nested model: AssessmentAnswer")
print("Supported question types: MCQ, open-ended")
print("Supported assessment types: adaptive_quiz, practice, assessment")
print("Purpose: Store assessment questions, student answers, and evaluation")
```
